```python
import json
import pandas as pd

file_path = '/content/Project5_KMeans_Patient_Segmentation.bin'

# Try to load the file as JSON
with open(file_path, 'r', encoding='utf-8') as f:
    try:
        data = json.load(f)
        print("Successfully loaded as JSON!")
        print(f"Loaded data type: {type(data)}")
        if isinstance(data, dict):
            print("Keys:", list(data.keys()))
            # If it contains tabular data, try displaying a sample
            for k, v in list(data.items())[:2]:
                print(f"\nKey: {k}, Type: {type(v)}")
                if isinstance(v, list) and len(v) > 0:
                    print("Sample element:", v[0])
        elif isinstance(data, list):
            print(f"Length of list: {len(data)}")
            print("Sample element:", data[0] if len(data) > 0 else "Empty list")
    except Exception as e:
        print("Failed to parse as JSON:", e)
        # Read the first few lines to manually inspect
        f.seek(0)
        print("First 200 characters of the file:")
        print(f.read(200))
```

```python
import json
import sys
from IPython import get_ipython

# Load the entire notebook
file_path = '/content/Project5_KMeans_Patient_Segmentation.bin'
with open(file_path, 'r', encoding='utf-8') as f:
    nb_data = json.load(f)

# Extract and execute all code cells sequentially
ipython = get_ipython()
code_cells = [cell for cell in nb_data['cells'] if cell['cell_type'] == 'code']

print(f"Executing {len(code_cells)} notebook cells sequentialy...")
for idx, cell in enumerate(code_cells):
    source_code = "".join(cell['source'])
    print(f"\n>>> Running Cell {idx + 1}/{len(code_cells)} (ID: {cell.get('id', 'N/A')}) ...")
    # Execute the code directly inside our current kernel
    ipython.run_cell(source_code)
```
